```python

%%writefile vetores.c
#include <stdio.h>

int main() {
    int vendas[5];
    int i;
    int soma = 0;

    // ETAPA 1: Ler os 5 valores
    for (i = 0; i < 5; i++) {
        printf("Digite as vendas do dia %d:\n", i + 1);
        scanf("%d", &vendas[i]);
    }

    // ETAPA 2: Calcular a soma
    for (i = 0; i < 5; i++) {
        soma = soma + vendas[i];
    }

    // ETAPA 3: Mostrar os valores
    printf("\nVendas registradas:\n");

    for (i = 0; i < 5; i++) {
        printf("Dia %d: %d vendas\n", i + 1, vendas[i]);
    }

    // ETAPA 4: Mostrar o total
    printf("\nTotal de vendas: %d\n", soma);

    return 0;
}
```
