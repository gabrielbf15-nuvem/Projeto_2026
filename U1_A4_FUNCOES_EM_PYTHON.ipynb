```python

# TRABALHO: SISTEMA DE GESTÃO DE NOTAS DE ALUNOS
# Disciplina: Linguagem de Programação

# Função para cadastrar as notas do aluno
def cadastrar_notas():
    notas = []

    # Pergunta quantas notas serão cadastradas
    quantidade = int(input("Quantas notas deseja cadastrar? "))

    # Repete a leitura de notas conforme a quantidade informada
    for i in range(quantidade):
        while True:
            nota = float(input(f"Digite a {i + 1}ª nota (0 a 10): "))

            # Verifica se a nota está dentro do intervalo permitido
            if 0 <= nota <= 10:
                notas.append(nota)
                break
            else:
                print("Nota inválida! Digite uma nota entre 0 e 10.")

    return notas


# Função para calcular a média das notas
def calcular_media(notas):
    return sum(notas) / len(notas)


# Função para determinar a situação do aluno
def verificar_situacao(media):
    if media >= 7:
        return "Aprovado"
    else:
        return "Reprovado"


# Função para exibir o relatório final
def exibir_relatorio(notas, media, situacao):
    print("\n===== RELATÓRIO FINAL DO ALUNO =====")
    print("Notas cadastradas:", notas)
    print(f"Média final: {media:.2f}")
    print("Situação:", situacao)
    print("====================================")


# Função principal que executa o sistema
def main():
    print("SISTEMA DE GESTÃO DE NOTAS")

    # Cadastra as notas
    notas = cadastrar_notas()

    # Calcula a média
    media = calcular_media(notas)

    # Verifica a situação do aluno
    situacao = verificar_situacao(media)

    # Apresenta o relatório
    exibir_relatorio(notas, media, situacao)


# Inicia o programa
if __name__ == "__main__":
    main()
```
