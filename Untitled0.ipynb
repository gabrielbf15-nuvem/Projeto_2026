```python
# ============================================================
# SISTEMA DE GESTÃO DE NOTAS DE ALUNOS
# Disciplina: Linguagem de Programação
# ============================================================

# Função para calcular a média das notas
def calcular_media(notas):
    """
    Recebe uma lista de notas e retorna a média aritmética.
    """
    if len(notas) == 0:
        return 0
    return sum(notas) / len(notas)


# Função para determinar a situação do aluno
def determinar_situacao(media):
    """
    Determina se o aluno está Aprovado ou Reprovado com base na média.
    Critério: Média >= 7.0 -> Aprovado | Média < 7.0 -> Reprovado
    """
    if media >= 7.0:
        return "Aprovado"
    else:
        return "Reprovado"


# Função principal para executar o programa
def executar_sistema():
    notas = []  # Lista para armazenar as notas inseridas

    print("=== SISTEMA DE CADASTRO DE NOTAS ===")
    print("Digite as notas dos alunos. Digite -1 quando terminar de inserir.\n")

    # Estrutura de repetição para leitura contínua das notas
    while True:
        try:
            entrada = float(input("Digite uma nota (ou -1 para encerrar): "))

            # Condição de saída do loop
            if entrada == -1:
                break

            # Validação para aceitar apenas notas entre 0 e 10
            if 0 <= entrada <= 10:
                notas.append(entrada)  # Adiciona a nota à lista
            else:
                print(
                    "Nota inválida! Por favor, insira um valor entre 0 e 10."
                )

        except ValueError:
            print("Entrada inválida! Por favor, digite um número válido.")

    # Verificação se alguma nota foi cadastrada
    if len(notas) == 0:
        print("\nNenhuma nota foi inserida.")
        return

    # Processamento dos dados através das funções
    media = calcular_media(notas)
    situacao = determinar_situacao(media)

    # Exibição do Relatório Final
    print("\n" + "=" * 35)
    print("         RELATÓRIO FINAL         ")
    print("=" * 35)
    print(f"Notas inseridas: {notas}")
    print(f"Média do aluno:  {media:.2f}")
    print(f"Situação final:  {situacao}")
    print("=" * 35)


# Executar o sistema
executar_sistema()+
```
