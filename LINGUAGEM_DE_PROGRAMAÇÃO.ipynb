```python
class pessoa:
  # Construtor: chamado quando criamos um objeto Pessoa
  def __init__(self, nome, idade, genero):
    self.nome = nome
    self.idade = idade
    self.genero = genero

  # Método que retorna uma saudação
  def cumprimentar(self):
    return f"Olá, meu nome é {self.nome}."

  # Método que almenta a idade em 1
  def aniversario(self):
    self.idade += 1
```

```python
pessoa1 = pessoa("João", 30, "masculino")
```

```python
print(pessoa1.cumprimentar())
print(f"Idade:  {pessoa1.idade}")
```

```python
# Chama o método idade aniversário para aumentar a idade
pessoa1.aniversario()
print(f"Nova idade: {pessoa1.idade}")
```

```python
# Herança - classe-pai Animal e duas classes-filhas
class Animal:
  def __init__(self, nome):
    self.nome = nome

  def fazer_barulho(self):
    pass   # método "vázio" que será subscrito pelas filhas
```

```python
class Cachorro(Animal):
  def fazer_barulho(self):
    return "latir"
```

```python
class Gato(Animal):
  def fazer_barulho(self):
    return "minar"
```

```python
# Criando e usando os objetos da classe-filhas
rex = Cachorro("Rex")
whiskers = Gato("Whiskers")
```

```python
print(f"{rex.nome} faz: {rex.fazer_barulho()}")
print(f"{whiskers.nome} faz: {whiskers.fazer_barulho()}")
```

```python
# classe Veiculo (classe-pai)
class Veiculo:
  def __init__(self, marca, modelo, ano):
    self.marca = marca
    self.modelo = modelo
    self.ano = ano
    self.velocidade = 0 # todos os veiculos começam parado

  def acelerar(self, incremento):
    self.velocidade += incremento

  def frear(self, decremento):
    self.velocidade -= decremento

  def status(self):
    return (f"Marca: {self.marca}, Modelo: {self.modelo}, ({self.ano})"
            f"Ano: {self.ano}, Velocidade: {self.velocidade} Km/h")
```

```python
# Classe-filha - Carro e Bicicleta
class Carro(Veiculo):
  def __init__(self, marca, modelo, ano, potencia):
    super().__init__(marca, modelo, ano)
    self.potencia = potencia

  # Sobscrevendo o método acelerar
  def acelerar(self, incremento):
    self.velocidade += incremento + self.potencia

class Bicicleta(Veiculo):
  def __init__(self, marca, modelo, ano, tipo):
    super().__init__(marca, modelo, ano)
    self.tipo = tipo

    # Vai subscrever o status tipo
    def status(self):
        return (f"Marca: {self.marca}, Modelo: {self.modelo} ",
                f"Ano: {self.ano}, tipo: {self.tipo}, "
                f"Velocidade, {self.velocidade} Km/h")
```

```python
# Testando os objetos
carro1 = Carro("Toyota", "Corolla", 2022, 150)
bicicleta1 = Bicicleta("Trek", "Mountanbike", 2021, "MTB'")

carro1.acelerar(50)
bicicleta1.frear(20)

print("Status do carro:")
print(carro1.status())

print("\nStatus da bicicleta:")
print(bicicleta1.status())
```

```python
class Produtos:
  def __init__(self, nome, preco, quantidade):
    self.nome = nome
    self.preco = preco
    self.quantidade = quantidade

def __init__(self):
    return(f"Nome: {self.nome}, Preço: {self.preco}, Quantidade: {self.quantidade}")
```

```python
catalogo = []
categorias = []
```

```python
def adicionar_produto(nome, preco, quantidade):
  produto = Produtos(nome, preco, quantidade)
  catalogo.append(produto)
  print(f"Produto '{nome}' adicionado ao catálogo.")
```

```python
def listar_produtos():
  print("\nCatálogo de Produtos:")
  for produto in catalogo:
    print(f"Nome: {produto.nome}, Preço: {produto.preco}, Quantidade: {produto.quantidade}")
```

```python
adicionar_produto("Camiseta", 29.99, 50)
adicionar_produto("Calça Jeans", 59.99, 30)
adicionar_produto("Tênis Esportivo", 89.99, 20)
adicionar_produto("Relógio", 79.99, 10)
adicionar_produto("Meia", 9.99, 100)
adicionar_produto("Sapato", 49.99, 40)

listar_produtos()
```

```python
import matplotlib.pyplot as plt

categorias_unicas = list(set(categorias))
categorias_unicas.sort()

contagem = [categorias.count(categoria) for categoria in categorias_unicas]

# O segundo argumento deve ser a lista de valores (contagem)
# O parâmetro de cor deve ser especificado como color='royalblue'
plt.bar(categorias_unicas, contagem, color='royalblue')
plt.xlabel('Categorias')
plt.ylabel('Quantidade')
plt.title('Distribuição de Produtos por Categoria')
plt.show()
```
