```python

%%writefile soma.c
#include <stdio.h>

int main() {
    int numero;
    int soma = 0;

    printf("Digite um numero inteiro (0 para encerrar):\n");
    scanf("%d", &numero);

    while (numero != 0) {
        soma = soma + numero;

        printf("Digite outro numero inteiro (0 para encerrar):\n");
        scanf("%d", &numero);
    }

    printf("Soma total: %d\n", soma);

    return 0;
}
```
