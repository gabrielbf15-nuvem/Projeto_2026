```python

%%writefile programa.c
#include <stdio.h>

int main() {
    int num1, num2, num3;

    printf("Digite o primeiro numero: ");
    scanf("%d", &num1);

    printf("Digite o segundo numero: ");
    scanf("%d", &num2);

    printf("Digite o terceiro numero: ");
    scanf("%d", &num3);

    // OPERADORES ARITMETICOS
    printf("\n=== OPERACOES ARITMETICAS ===\n");

    printf("Soma: %d\n", num1 + num2 + num3);
    printf("Subtracao: %d\n", num1 - num2 - num3);
    printf("Multiplicacao: %d\n", num1 * num2 * num3);

    if (num2 != 0 && num3 != 0) {
        float divisao = (float)num1 / num2 / num3;
        printf("Divisao: %.2f\n", divisao);
    } else {
        printf("Divisao impossivel: divisao por zero.\n");
    }

    // OPERADORES RELACIONAIS
    printf("\n=== OPERADORES RELACIONAIS ===\n");

    if (num1 > num2) {
        printf("O primeiro numero e maior que o segundo.\n");
    } else {
        printf("O primeiro numero NAO e maior que o segundo.\n");
    }

    if (num2 < num3) {
        printf("O segundo numero e menor que o terceiro.\n");
    } else {
        printf("O segundo numero NAO e menor que o terceiro.\n");
    }

    // OPERADORES LOGICOS
    printf("\n=== OPERADORES LOGICOS ===\n");

    if (num1 > 0 && num2 % 2 == 0) {
        printf("Condicao verdadeira: primeiro positivo e segundo par.\n");
    } else {
        printf("Condicao falsa: os requisitos nao foram atendidos.\n");
    }

    return 0;
}
```
