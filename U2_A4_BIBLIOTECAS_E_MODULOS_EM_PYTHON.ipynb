```python

# TRABALHO: SISTEMA DE GERENCIAMENTO DE LIVROS
# Disciplina: Linguagem de Programação
# Unidade 2 - Bibliotecas e Módulos em Python

# Importa a biblioteca usada para criar gráficos
import matplotlib.pyplot as plt


# Classe que representa um livro
class Livro:
    def __init__(self, titulo, autor, genero, quantidade):
        self.titulo = titulo
        self.autor = autor
        self.genero = genero
        self.quantidade = quantidade


# Lista que armazenará os livros cadastrados
biblioteca = []


# Função para cadastrar um livro
def cadastrar_livro():
    print("\n--- CADASTRO DE LIVRO ---")

    titulo = input("Digite o título: ").strip()
    autor = input("Digite o autor: ").strip()
    genero = input("Digite o gênero: ").strip()

    # Verifica se os campos de texto foram preenchidos
    if not titulo or not autor or not genero:
        print("Erro: todos os campos devem ser preenchidos.")
        return

    # Solicita uma quantidade inteira não negativa
    try:
        quantidade = int(input("Quantidade disponível: "))
        if quantidade < 0:
            print("A quantidade não pode ser negativa.")
            return
    except ValueError:
        print("Digite uma quantidade inteira válida.")
        return

    # Cria o objeto e guarda na lista
    livro = Livro(titulo, autor, genero, quantidade)
    biblioteca.append(livro)

    print("Livro cadastrado com sucesso!")


# Função para listar todos os livros
def listar_livros():
    print("\n--- LIVROS CADASTRADOS ---")

    if not biblioteca:
        print("Nenhum livro cadastrado.")
        return

    for livro in biblioteca:
        print(f"Título: {livro.titulo}")
        print(f"Autor: {livro.autor}")
        print(f"Gênero: {livro.genero}")
        print(f"Quantidade disponível: {livro.quantidade}")
        print("-" * 30)


# Função para buscar um livro pelo título
def buscar_livro():
    titulo_busca = input(
        "\nDigite o título do livro que deseja buscar: "
    ).strip().lower()

    encontrados = [
        livro for livro in biblioteca
        if titulo_busca in livro.titulo.lower()
    ]

    if encontrados:
        print("\n--- RESULTADO DA BUSCA ---")
        for livro in encontrados:
            print(f"Título: {livro.titulo}")
            print(f"Autor: {livro.autor}")
            print(f"Gênero: {livro.genero}")
            print(f"Quantidade disponível: {livro.quantidade}")
    else:
        print("Livro não encontrado.")


# Função para gerar um gráfico de livros por gênero
def gerar_grafico():
    if not biblioteca:
        print("Cadastre pelo menos um livro para gerar o gráfico.")
        return

    # Soma a quantidade disponível de livros de cada gênero
    quantidades = {}

    for livro in biblioteca:
        if livro.genero in quantidades:
            quantidades[livro.genero] += livro.quantidade
        else:
            quantidades[livro.genero] = livro.quantidade

    # Monta o gráfico
    plt.figure(figsize=(8, 5))
    plt.bar(quantidades.keys(), quantidades.values())
    plt.title("Quantidade de Livros por Gênero")
    plt.xlabel("Gênero")
    plt.ylabel("Quantidade de livros")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.show()


# Adiciona livros de exemplo para facilitar o primeiro teste
biblioteca.append(
    Livro("Dom Casmurro", "Machado de Assis", "Romance", 3)
)
biblioteca.append(
    Livro("O Hobbit", "J. R. R. Tolkien", "Fantasia", 5)
)
biblioteca.append(
    Livro("A Revolução dos Bichos", "George Orwell", "Ficção", 2)
)
biblioteca.append(
    Livro("Orgulho e Preconceito", "Jane Austen", "Romance", 4)
)


# Menu principal do sistema
def menu():
    while True:
        print("\n===== BIBLIOTECA =====")
        print("1 - Cadastrar livro")
        print("2 - Listar livros")
        print("3 - Buscar livro pelo título")
        print("4 - Gerar gráfico por gênero")
        print("0 - Sair")

        opcao = input("Escolha uma opção: ").strip()

        if opcao == "1":
            cadastrar_livro()
        elif opcao == "2":
            listar_livros()
        elif opcao == "3":
            buscar_livro()
        elif opcao == "4":
            gerar_grafico()
        elif opcao == "0":
            print("Sistema encerrado.")
            break
        else:
            print("Opção inválida. Tente novamente.")


# Inicia o sistema
menu()
```
